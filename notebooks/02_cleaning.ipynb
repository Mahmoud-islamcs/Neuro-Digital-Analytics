# Brain Rot Project - Data Cleaning

## 0. Import Libraries and Load Raw Data

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)

# Load the original (raw) csv files
path = r"F:\Data Analysis\DEPI\Final Final Project\Depi-Data-Analysis-Project(github project structure)\raw_data"

fact = pd.read_csv(f"{path}/Fact_User_Activity.csv")
users = pd.read_csv(f"{path}/Dim_Users.csv")
habits = pd.read_csv(f"{path}/Dim_Habits.csv")
mentalstate = pd.read_csv(f"{path}/Dim_MentalState.csv")
date = pd.read_csv(f"{path}/Dim_Date.csv")

print("All files loaded successfully!")
```

## 1. Cleaning Fact_User_Activity

```python
# Step 1: check missing values
print("Missing values before cleaning:")
print(fact.isnull().sum())
```

```python
# Step 2: check duplicated rows
print("Duplicated rows :", fact.duplicated().sum())
```

```python
# Step 3: check for impossible values in the important numeric columns
# Importante !!!! (a value is "impossible" if it is negative, when negative does not make sense)

print("Minimum Study_Hours:", fact['Study_Hours'].min())
print("Minimum Coffee_Consumed_Per_Day:", fact['Coffee_Consumed_Per_Day'].min())
print("Minimum Smoking_Breaks_Count:", fact['Smoking_Breaks_Count'].min())
print("Minimum Total_Reels_Watched:", fact['Total_Reels_Watched'].min()) 
print("Minimum Focus_Sessions_Count:", fact['Focus_Sessions_Count'].min())
print("Minimum Brainrot_Exposure_Score:", fact['Brainrot_Exposure_Score'].min())
print("Minimum Wellbeing_Score:", fact['Wellbeing_Score'].min())
```

```python
# Step 4: check data types are correct
print(fact.dtypes)
```

```python
fact_clean = fact.copy()
print("Fact_User_Activity is now clean. Shape:", fact_clean.shape)
```

## 2. Cleaning Dim_Users

```python
print("Missing values before cleaning:")
print(users.isnull().sum())
```

```python
print("Duplicated rows :", users.duplicated().sum())
```

```python
print("Minimum Age:", users['Age'].min())
print("Maximum Age:", users['Age'].max())
```

```python
users_clean = users.copy()
print(users_clean.shape)
```

## 3. Cleaning Dim_Habits

```python
habits
```

```python
print("Missing values before cleaning:")
print(habits.isnull().sum())
```

```python
habits['Coffee_Level'] = habits['Coffee_Level'].fillna('Unknown')
print("Missing values after cleaning:")
print(habits.isnull().sum())
```

```python
print("Duplicated rows:", habits.duplicated().sum())
habits_clean = habits.copy()
habits_clean
```

## 4. Cleaning Dim_MentalState

```python
print("Missing values:")
print(mentalstate.isnull().sum())
print("Duplicated rows:", mentalstate.duplicated().sum())
mentalstate_clean = mentalstate.copy()
mentalstate_clean
```

## 5. Cleaning Dim_Date

This table has one row per calendar day. We convert `FullDate` from text to a real date type.

```python
print("Missing values before cleaning:")
print(date.isnull().sum())
print("Duplicated rows:", date.duplicated().sum())
```

```python
print("Data type before conversion:", date['FullDate'].dtype)
date['FullDate'] = pd.to_datetime(date['FullDate'])
print(date['FullDate'].dtype)
```

```python
date_clean = date.copy()
print(date_clean.shape)
date_clean.head()
```

## 6. Checking Relationships Between Tables (Foreign Keys)

```python
users_not_found = fact_clean[~fact_clean['UserKey'].isin(users_clean['UserKey'])]
print("Number of fact rows with a UserKey that does NOT exist in Dim_Users:", len(users_not_found))
```

```python
dates_not_found = fact_clean[~fact_clean['DateKey'].isin(date_clean['DateKey'])]
print("Number of fact rows with a DateKey that does NOT exist in Dim_Date:", len(dates_not_found))
```

```python
habits_not_found = fact_clean[~fact_clean['HabitKey'].isin(habits_clean['HabitKey'])]
print("Number of fact rows with a HabitKey that does NOT exist in Dim_Habits:", len(habits_not_found))
```

```python
states_not_found = fact_clean[~fact_clean['StateKey'].isin(mentalstate_clean['StateKey'])]
print("Number of fact rows with a StateKey that does NOT exist in Dim_MentalState:", len(states_not_found))
```

## 7. Save the Cleaned CSV Files

```python
fact_clean.to_csv('Fact_User_Activity.csv', index=False)
users_clean.to_csv('Dim_Users.csv', index=False)
habits_clean.to_csv('Dim_Habits.csv', index=False)
mentalstate_clean.to_csv('Dim_MentalState.csv', index=False)
date_clean.to_csv('Dim_Date.csv', index=False)
```

## 8. Full Join - Building the Final Merged Dataset

```python
# Start from the fact table and add the users information
final_df = fact_clean.merge(users_clean, on='UserKey', how='left')
print("Shape after merging with Dim_Users:", final_df.shape)
```

```python
final_df = final_df.merge(date_clean, on='DateKey', how='left')
print("Shape after merging with Dim_Date:", final_df.shape)
```

```python
final_df = final_df.merge(habits_clean, on='HabitKey', how='left')
print("Shape after merging with Dim_Habits:", final_df.shape)
```

```python
final_df = final_df.merge(mentalstate_clean, on='StateKey', how='left')
print("Shape after merging with Dim_MentalState:", final_df.shape)
```

```python
final_df.to_csv('BrainRot_Final_Dataset.csv', index=False)
```
