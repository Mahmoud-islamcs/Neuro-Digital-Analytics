```python
import random
from faker import Faker
from google.colab import files # for loading the data
```

```python
fake = Faker()
num_users = 20000
file_name = "Brain_Rot_Analysis_Data.sql"
file_name = "Brain_Rot_New_Tables_Data.sql"
occupations = ['Student', 'Software Engineer', 'Content Creator', 'Doctor', 'Marketing Specialist', 'Unemployed', 'Call Center Agent']
platforms = ['TikTok', 'Instagram Reels', 'YouTube Shorts', 'Facebook', 'YouTube', 'Twitter']
```

```python
with open(file_name, 'w', encoding='utf-8') as f:
    f.write("USE BrainRotAnalysis;\nGO\n\n")
    for user_id in range(1, num_users + 1):
        age = random.randint(15, 45)
        gender = random.choice(['Male', 'Female'])
        occ = random.choice(occupations)
        sleep = round(random.uniform(4, 7), 1) if age < 25 else round(random.uniform(6, 9), 1)
        f.write(f"INSERT INTO Users (age, gender, occupation, avg_sleep_hours) VALUES ({age}, '{gender}', '{occ}', {sleep});\n")
        is_heavy = random.random() > 0.3
        if is_heavy:
            platform, content, usage, scroll = random.choice(['TikTok', 'Instagram Reels']), 'Short-form', random.randint(240, 600), random.randint(7, 10)
            attn, react, anxiety, boredom, mood = random.randint(5, 40), random.randint(500, 900), random.randint(7, 10), random.randint(1, 3), random.choice(['Low', 'Very Low'])
        else:
            platform, content, usage, scroll = random.choice(['YouTube', 'Facebook']), random.choice(['Long-form', 'Static-Posts']), random.randint(30, 150), random.randint(1, 5)
            attn, react, anxiety, boredom, mood = random.randint(60, 95), random.randint(200, 450), random.randint(1, 5), random.randint(6, 10), random.choice(['High', 'Neutral'])
        f.write(f"INSERT INTO Digital_Consumption (user_id, platform_name, content_format, daily_usage_minutes, scrolling_speed_index) VALUES ({user_id}, '{platform}', '{content}', {usage}, {scroll});\n")
        f.write(f"INSERT INTO Cognitive_Metrics (user_id, attention_span_score, memory_recall_accuracy, reaction_time_ms, test_date) VALUES ({user_id}, {attn}, {round(random.uniform(30, 99), 2)}, {react}, '{fake.date_between(start_date='-1y', end_date='today')}');\n")
        f.write(f"INSERT INTO Psychological_State (user_id, anxiety_level, loneliness_score, boredom_tolerance, mood_rating) VALUES ({user_id}, {anxiety}, {random.randint(1,10)}, {boredom}, '{mood}');\n")
```

```python
print(f"File '{file_name}' created successfully.")
```

```python
files.download(file_name)
```

```python
with open(file_name, 'w', encoding='utf-8') as f:
    f.write("USE BrainRotAnalysis;\nGO\n\n")
    for user_id in range(1, num_users + 1):
        is_heavy = random.random() > 0.4

        # 1. Physical_Health
        eye_strain = random.randint(7, 10) if is_heavy else random.randint(1, 5)
        neck_pain = 1 if is_heavy else 0
        steps = random.randint(1000, 3000) if is_heavy else random.randint(5000, 10000)
        f.write(f"INSERT INTO Physical_Health (user_id, eye_strain_level, neck_back_pain, daily_steps) VALUES ({user_id}, {eye_strain}, {neck_pain}, {steps});\n")

        # 2. Performance_Metrics
        gpa = round(random.uniform(1.5, 2.5), 2) if is_heavy else round(random.uniform(3.0, 4.0), 2)
        deadlines = random.randint(5, 15) if is_heavy else random.randint(0, 2)
        work_focus = random.randint(1, 4) if is_heavy else random.randint(7, 10)
        f.write(f"INSERT INTO Performance_Metrics (user_id, gpa_or_rating, missed_deadlines_count, focus_during_work_score) VALUES ({user_id}, {gpa}, {deadlines}, {work_focus});\n")

        # 3. Social_Impact
        social_hours = round(random.uniform(0.5, 2.0), 1) if is_heavy else round(random.uniform(3.0, 6.0), 1)
        social_anxiety = random.randint(6, 10) if is_heavy else random.randint(1, 4)
        f.write(f"INSERT INTO Social_Impact (user_id, face_to_face_interaction_hours, social_anxiety_level, family_conflicts_score) VALUES ({user_id}, {social_hours}, {social_anxiety}, {random.randint(1,10)});\n")

        # 4. Sleep_Details
        latency = random.randint(45, 120) if is_heavy else random.randint(5, 20)
        phone_before = random.randint(60, 180) if is_heavy else random.randint(0, 30)
        f.write(f"INSERT INTO Sleep_Details (user_id, latency_to_sleep_minutes, night_awakenings, phone_before_sleep_minutes) VALUES ({user_id}, {latency}, {random.randint(0,5)}, {phone_before});\n")

        # 5. Content_Interests
        category = random.choice(['BrainRot-Core', 'Memes', 'ASMR']) if is_heavy else random.choice(['Education', 'News', 'Documentary'])
        is_edu = 0 if is_heavy else 1
        video_count = random.randint(100, 500) if is_heavy else random.randint(5, 20)
        f.write(f"INSERT INTO Content_Interests (user_id, primary_category, is_educational_content, daily_video_count) VALUES ({user_id}, '{category}', {is_edu}, {video_count});\n")

        # 6. Financial_Metrics
        spending = round(random.uniform(50, 500), 2) if is_heavy else 0.00
        impulse = random.randint(7, 10) if is_heavy else random.randint(1, 4)
        f.write(f"INSERT INTO Financial_Metrics (user_id, in_app_purchases_usd, impulse_buying_frequency) VALUES ({user_id}, {spending}, {impulse});\n")
    f.write("GO\n")
```

```python
print(f"File '{file_name}' created successfully with logically linked data.")
```
